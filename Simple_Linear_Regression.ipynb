```python
#  Imports Libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
#Load and print first 5 rows

# read csv file
data = pd.read_csv("mental_health.csv")

data.head()
```

```python
# Create simple linear model for predict stress according to one feature sleep hours

# declare independent and dependent values
X = data[['Sleep_Hours_Night']]      
y = data['Work_Stress_Level']       


# create linear regression model
model = LinearRegression()


# train the model
model.fit(X, y)


# predict stress for 5 hours sleep
predicted_stress = model.predict([[5]])

print("Predicted Stress Level for 5 hours sleep:", predicted_stress[0])


# predict values for regression line
y_pred = model.predict(X)
print("R2 Score:", r2_score(y, y_pred))


# plot graph
plt.scatter(X, y)          
plt.plot(X, y_pred)        
plt.scatter(5, predicted_stress)   

plt.xlabel("Sleep Hours Night")
plt.ylabel("Work Stress Level")
plt.title("Simple Linear Regression (Sleep vs Stress)")
plt.show()
```
