```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
# Load data
data = pd.read_csv("mental_health.csv")
```

```python
#create logistic linear regression model for predict mental health issue based on 3 4 inputs


# Inputs (multiple)
X = data[['Sleep_Hours_Night',
          'Work_Hours_Per_Week',
          'Screen_Time_Hours_Day',
          'Social_Media_Hours_Day',
          'Work_Stress_Level']]

# Output (0 or 1)
y = data['Has_Mental_Health_Issue']

# Create model
model = LogisticRegression(max_iter=1000)

X_train, X_test, y_train, y_test = train_test_split(X, y)

# Train
model.fit(X_train, y_train)

# Predict whole dataset
y_pred = model.predict(X_test)

# Print predicted values
print("Predicted:", y_pred)


# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))


# Example prediction (single person)
pred = model.predict([[6, 45, 5, 3, 7]])
print("New Prediction (0=No, 1=Yes) that a person has mental health issue:", pred[0])

```
