```python
#Age Distribution using histogram

plt.figure()
plt.hist(df['Age'], bins=20, color='skyblue', edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Gender Distribution Bar Chart

plt.figure()
df['Gender'].value_counts().plot(kind='bar', color='mediumseagreen')
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

```python
#Mental Health Issue Count

plt.figure()
df['Has_Mental_Health_Issue'].value_counts().plot(kind='bar', color=['purple','gold'])
plt.title("Mental Health Issue Count")
plt.xlabel("Issue")
plt.ylabel("Count")
plt.show()
```

```python
#Employment Status

counts = df['Employment_Status'].value_counts()

plt.pie(counts, labels=counts.index, autopct='%1.1f%%')
plt.title("Employment Status")
plt.show()
```

```python
#Mental Health Treatment

counts = df['Ever_Sought_Treatment'].value_counts()

plt.pie(counts, labels=counts.index, autopct='%1.1f%%')
plt.title("Mental Health Treatment Taken")
plt.show()
```
